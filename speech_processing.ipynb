1. Polarize a matrix

```python
import numpy as np
import random
from numpy import *


def polarize_matrix(matrix: ndarray)-> ndarray:
    #this function converts all elements of an arbitrary sized matrix to -1 if the element 
    # is negative, in other case, to 1

    return where(matrix < 0, -1, 1)

random_matrix = random.rand(5, 10) - 0.5
polarized_random_matrix = polarize_matrix(random_matrix)

print(random_matrix)
print(polarized_random_matrix)
```

2. Simulation of coin tosses


```python
def coin_toss_probability(number_of_trials: int)-> float:
    #function that simulates a number of consecutive coin tosses and returns the relative 
    #frequency of having tossed "heads"
    random_trials=random.rand(number_of_trials)
    number_of_heads=np.count_nonzero(random_trials>=0.5)
    return number_of_heads / number_of_trials

print(coin_toss_probability(500))
```

3. Pascal's Triangle

```python
def pascals_triangle(number_of_rows: int) -> ndarray:
    #this function generates a pascals triangle
    number_of_columns=number_of_rows
    result=zeros((number_of_rows, number_of_columns), dtype=int)

    for (i, j) in ndindex(number_of_rows, number_of_columns):
        if j>i:
            continue
        if j==0 or i==j:
            result[i, j]=1
        else:
            result[i,j]=result[i-1, j-1]+result[i-1,j]
    return result

#shifting each column down by the number of elements equal to the column index
size=50
pasc_matrix=pascals_triangle(size)
print(pasc_matrix)

shifted_matrix=np.zeros((size, size), dtype=np.int64)

for j in range(size):
    shifted_matrix[j:, j] = pasc_matrix[:size - j, j]

row_sums_shifted = shifted_matrix.sum(axis=1)
print("\nRow sums (Shifted method):")
print(row_sums_shifted)

#other method: using numpy.diagonal()

p_matrix_flipped=np.flipud(pasc_matrix)

row_sums_diag=np.array([np.sum(np.diagonal(p_matrix_flipped, offset=-(size-1-k)))
        for k in range (size)                    
])

print("\nRow sums (Diagonal method):")
print(row_sums_diag)

print("\nThe name of the result sequence is Fibonacci sequence.\n")

#golden ratio

import scipy.constants as const

phi=const.golden

ratios=row_sums_shifted[1:] / row_sums_shifted[:-1]

golden_prefix = f"{phi:.10f}"[:6]

for count, ratio in enumerate(ratios, start=1):
    ratio_prefix = f"{ratio:.10f}"[:6]
    if ratio_prefix == golden_prefix:
        print(f"After {count} ratios, we get a ratio ({ratio:.8f}) that matches the golden ratio by the first 5 digits ('{golden_prefix}').")
        break
```
